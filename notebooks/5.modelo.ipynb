## 2. Entrenando el Primer Modelo

### 2.1 Modelo Inicial (Base)

```python
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score, median_absolute_error

# --- Entrenamiento del modelo ---
rfr_inicial = RandomForestRegressor(
    n_estimators=100,
    max_depth=10,
    random_state=42,
    n_jobs=-1,
    oob_score=True
)
rfr_inicial.fit(x_train, y_train)

# --- Predicciones ---
y_pred_train = rfr_inicial.predict(x_train)

# --- Métricas ---
mse_inicial = mean_squared_error(y_train, y_pred_train)
mae_inicial = mean_absolute_error(y_train, y_pred_train)
rmse_inicial = np.sqrt(mse_inicial)
medae_inicial = median_absolute_error(y_train, y_pred_train)
r2_inicial = r2_score(y_train, y_pred_train)
dif_oob_r2_inicial =  r2_inicial - rfr_inicial.oob_score_


print("=== Métricas de entrenamiento ===")
print(f"MAE  : {mae_inicial:.2f} puntos → en promedio, el modelo se equivoca ±{mae_inicial:.2f} puntos respecto al puntaje real del ICFES.")
print(f"RMSE : {rmse_inicial:.2f} puntos → refleja el impacto de los errores grandes; si el modelo falla mucho en casos extremos, este valor aumenta.")
print(f"MedAE: {medae_inicial:.2f} puntos → la mitad de las predicciones cometen menos de {medae_inicial:.2f} puntos de error.")
print(f"R²   : {r2_inicial:.4f} → el modelo explica el {r2_inicial*100:.1f}% de la variabilidad total del puntaje.")
print(f"OOB score: {rfr_inicial.oob_score_:.4f} → estimación del rendimiento del modelo en datos no vistos (validación interna del Random Forest).")

print(f"Diferencia R² - OOB: {dif_oob_r2_inicial:.4f} → "
      f"{'indica sobreajuste (el modelo memoriza demasiado)' if dif_oob_r2_inicial > 0 else 'no muestra señales de sobreajuste'} "
      f"porque el rendimiento real estimado es {abs(dif_oob_r2_inicial):.4f} puntos {'menor' if dif_oob_r2_inicial > 0 else 'mayor'} que el obtenido en entrenamiento.")
```

### 2.2 Búsqueda de Hiperparámetros

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_absolute_error


# ==============================
# BÚSQUEDA INDIVIDUAL
# ==============================

N_ESTIMADORES = np.arange(10, 301, 10)
mae_scores = []

for n_arboles in N_ESTIMADORES:
    print(f"Entrenando con {n_arboles} árboles...")

    bosque = RandomForestRegressor(
        n_estimators=n_arboles,
        random_state=42,
        oob_score=True,
        bootstrap=True,
        n_jobs=-1
    )

    bosque.fit(x_train, y_train)

    # Predicciones OOB
    oob_pred = bosque.oob_prediction_

    # MAE
    mae = mean_absolute_error(y_train, oob_pred)
    mae_scores.append(mae)


# ==============================
# GRAFICAR
# ==============================

plt.figure(figsize=(12, 7))
plt.style.use("seaborn-v0_8-whitegrid")

plt.plot(N_ESTIMADORES, mae_scores, marker="o", linewidth=2)
plt.xlabel("Número de árboles", fontsize=12)
plt.ylabel("MAE OOB", fontsize=12)
plt.title("Búsqueda Individual de n_estimators (MAE OOB)", fontsize=14, weight="bold")

# Punto óptimo (menor MAE)
mejor_mae = min(mae_scores)
mejor_n = N_ESTIMADORES[np.argmin(mae_scores)]

# Línea vertical
plt.axvline(mejor_n, color="gray", linestyle="--", alpha=0.8)

# Punto resaltado
plt.scatter(mejor_n, mejor_mae, color="red", s=100, zorder=5)

# Flecha + texto
plt.annotate(
    f"Mejor MAE = {mejor_mae:.4f}\n{mejor_n} árboles",
    xy=(mejor_n, mejor_mae),
    xytext=(mejor_n + 20, mejor_mae + (0.03 * mejor_mae)),
    arrowprops=dict(arrowstyle="->", color="black", lw=2),
    fontsize=11,
    bbox=dict(facecolor="white", edgecolor="black", alpha=0.8)
)

plt.tight_layout()
plt.show()
```

### 2.4 Comparando los modelos inicial vs afinado

```python
def reporte_comparacion_modelos(
    mae_inicial, rmse_inicial, medae_inicial, r2_inicial, dif_oob_r2_inicial, oob_inicial,
    mae, rmse, medae, r2, dif_oob_r2, oob_final
):

    print("\n==============================================")
    print("   REPORTE FINAL DE COMPARACIÓN DE MODELOS")
    print("==============================================\n")

    print("1. Comparación métrica por métrica\n")

    print(f"{'Métrica':<20} {'Inicial':<15} {'Final':<15} {'Mejor'}")
    print("-"*60)
    print(f"{'MAE':<20} {mae_inicial:<15.3f} {mae:<15.3f} "
          f"{'Final' if mae < mae_inicial else 'Inicial'}")
    print(f"{'RMSE':<20} {rmse_inicial:<15.3f} {rmse:<15.3f} "
          f"{'Final' if rmse < rmse_inicial else 'Inicial'}")
    print(f"{'MedAE':<20} {medae_inicial:<15.3f} {medae:<15.3f} "
          f"{'Final' if medae < medae_inicial else 'Inicial'}")
    print(f"{'R² (train)':<20} {r2_inicial:<15.4f} {r2:<15.4f} "
          f"{'Final' if r2 > r2_inicial else 'Inicial'}")
    print(f"{'OOB Score':<20} {oob_inicial:<15.4f} {oob_final:<15.4f} "
          f"{'Final' if oob_final > oob_inicial else 'Inicial'}")
    print(f"{'R² - OOB':<20} {dif_oob_r2_inicial:<15.4f} {dif_oob_r2:<15.4f} "
          f"{'Final' if abs(dif_oob_r2) < abs(dif_oob_r2_inicial) else 'Inicial'}")

    print("\n2. Interpretación del sobreajuste\n")

    def interpretacion_sobreajuste(dif):
        if dif < 0.05:
            return "Bajo sobreajuste"
        elif dif < 0.15:
            return "Sobreajuste moderado"
        else:
            return "Sobreajuste alto"

    print(f"Modelo inicial: {interpretacion_sobreajuste(dif_oob_r2_inicial)}")
    print(f"Modelo final  : {interpretacion_sobreajuste(dif_oob_r2)}")

    print("\n3. ¿Cuál modelo generaliza mejor?")
    print(f"→ {'Modelo final' if oob_final > oob_inicial else 'Modelo inicial'} "
          "porque tiene un OOB score mayor.")

    print("\n4. Conclusión final")
    if rmse < rmse_inicial and oob_final > oob_inicial:
        print("El modelo final es superior en error y generalización.")
    else:
        print("El modelo inicial sigue siendo competitivo según algunas métricas.")

    print("\n==============================================\n")

reporte_comparacion_modelos(
    mae_inicial, rmse_inicial, medae_inicial, r2_inicial, dif_oob_r2_inicial, rfr.oob_score_,
    mae, rmse, medae, r2, dif_oob_r2, rfr.oob_score_
)
```

### 3.3 Comparando los 3 Modelos

```python
def reporte_comparacion_3_modelos(
    # Modelo inicial
    mae_ini, rmse_ini, medae_ini, r2_ini, oob_ini, dif_ini,
    # Modelo intermedio
    mae_mid, rmse_mid, medae_mid, r2_mid, oob_mid, dif_mid,
    # Modelo optimizado
    mae_opt, rmse_opt, medae_opt, r2_opt, oob_opt, dif_opt
):

    print("\n==============================================")
    print("   REPORTE COMPARATIVO DE TRES MODELOS")
    print("==============================================\n")

    # Tabla de comparación
    print(f"{'Métrica':<14} {'Inicial':<12} {'Intermedio':<12} {'Optimizado':<12} {'Mejor'}")
    print("-"*70)

    def mejor(valor_ini, valor_mid, valor_opt, es_error=True):
        """Devuelve texto indicando qué modelo es mejor."""
        if es_error:
            m = min(valor_ini, valor_mid, valor_opt)
        else:
            m = max(valor_ini, valor_mid, valor_opt)

        if m == valor_ini:   return "Inicial"
        if m == valor_mid:   return "Intermedio"
        return "Optimizado"

    print(f"{'MAE':<14} {mae_ini:<12.3f} {mae_mid:<12.3f} {mae_opt:<12.3f} {mejor(mae_ini, mae_mid, mae_opt)}")
    print(f"{'RMSE':<14} {rmse_ini:<12.3f} {rmse_mid:<12.3f} {rmse_opt:<12.3f} {mejor(rmse_ini, rmse_mid, rmse_opt)}")
    print(f"{'MedAE':<14} {medae_ini:<12.3f} {medae_mid:<12.3f} {medae_opt:<12.3f} {mejor(medae_ini, medae_mid, medae_opt)}")
    print(f"{'R² (train)':<14} {r2_ini:<12.4f} {r2_mid:<12.4f} {r2_opt:<12.4f} {mejor(r2_ini, r2_mid, r2_opt, es_error=False)}")
    print(f"{'OOB Score':<14} {oob_ini:<12.4f} {oob_mid:<12.4f} {oob_opt:<12.4f} {mejor(oob_ini, oob_mid, oob_opt, es_error=False)}")
    print(f"{'R² - OOB':<14} {dif_ini:<12.4f} {dif_mid:<12.4f} {dif_opt:<12.4f} {mejor(abs(dif_ini), abs(dif_mid), abs(dif_opt))}")

    # Interpretación de sobreajuste
    def interpretacion(dif):
        if abs(dif) < 0.05:
            return "Bajo sobreajuste"
        elif abs(dif) < 0.15:
            return "Sobreajuste moderado"
        else:
            return "Sobreajuste alto"

    print("\n--- SOBREAJUSTE POR MODELO ---")
    print(f"Inicial    : {interpretacion(dif_ini)}")
    print(f"Intermedio : {interpretacion(dif_mid)}")
    print(f"Optimizado : {interpretacion(dif_opt)}")

    # Mejor modelo en general (por MAE + OOB)
    print("\n--- MEJOR MODELO GLOBAL ---")
    puntaje_ini = mae_ini - oob_ini
    puntaje_mid = mae_mid - oob_mid
    puntaje_opt = mae_opt - oob_opt

    mejor_global = mejor(puntaje_ini, puntaje_mid, puntaje_opt)
    print(f"→ El modelo {mejor_global} es el mejor considerando MAE y OOB Score.")

    print("\n==============================================\n")

reporte_comparacion_3_modelos(

    # Modelo inicial
    mae_inicial, rmse_inicial, medae_inicial, r2_inicial, rfr_inicial.oob_score_, dif_oob_r2_inicial,

    # Modelo intermedio (tu modelo rfr)
    mae, rmse, medae, r2, rfr.oob_score_, dif_oob_r2,

    # Modelo optimizado
    mae_opt, rmse_opt, medae_opt, r2_opt, rfr_opt.oob_score_, dif_oob_r2_opt
)
```

## 4. Análizando el error

### 4.1 Error Contextualizado (Error comparado con la variación natural de los datos)

```python
# =============================================================================
# PRIMER ANÁLISIS: MÉTRICAS CONTEXTUALIZADAS (MAE)
# =============================================================================
print("="*60)
print("1. ANÁLISIS DE MÉTRICAS CONTEXTUALIZADAS (MAE)")
print("="*60)

# Obtener estadísticas descriptivas del puntaje ICFES
puntaje_stats = y_train.describe()
print("\nEstadísticas descriptivas del puntaje ICFES:")
print(puntaje_stats)

# Calcular desviación estándar
std_puntajes = y_train.std()
print(f"\nDesviación estándar de los puntajes: {std_puntajes:.2f}")

# Tu MAE actual
mae_actual = mae_opt

# Calcular MAE como proporción de la desviación estándar
mae_normalized = mae_actual / std_puntajes
print(f"MAE como proporción de la desviación estándar: {mae_normalized:.3f}")
print(f"Esto significa que tu error típico es de {mae_normalized:.1f} desviaciones estándar")

# Interpretación contextualizada
rango_real = y_train.max() - y_train.min()
error_porcentual_real = (mae_actual / rango_real) * 100
print(f"\nRango real de puntajes en tus datos: {rango_real:.1f} puntos")
print(f"MAE como % del rango real: {error_porcentual_real:.1f}%")

# Crear visualización de la distribución de puntajes
plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
plt.hist(y_train, bins=30, alpha=0.7, color='skyblue', edgecolor='black')

plt.axvline(y_train.mean(), color='red', linestyle='--', 
           label=f'Media: {y_train.mean():.1f}')

plt.axvline(y_train.mean() + mae_actual, color='orange', linestyle='--',
           label=f'Media + MAE: {y_train.mean() + mae_actual:.1f}')

plt.axvline(y_train.mean() - mae_actual, color='orange', linestyle='--',
           label=f'Media - MAE: {y_train.mean() - mae_actual:.1f}')

plt.xlabel('Puntaje ICFES')
plt.ylabel('Frecuencia')
plt.title('Distribución de Puntajes ICFES\ncon Bandas de Error MAE')
plt.legend()
plt.grid(True, alpha=0.3)

plt.subplot(1, 2, 2)
plt.boxplot(y_train)
plt.ylabel('Puntaje ICFES')
plt.title('Boxplot de Puntajes ICFES')
plt.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```

### 4.2 Examen de Residuos

```python
from scipy import stats
# =============================================================================
# SEGUNDO ANÁLISIS: EXAMEN DE RESIDUOS
# =============================================================================
print("\n" + "="*60)
print("2. ANÁLISIS DE RESIDUOS DEL MODELO")
print("="*60)

# Generar predicciones con el modelo entrenado
y_pred_train_final = rfr_opt.predict(x_train_optimizado)

# Calcular residuos reales
residuos = y_train - y_pred_train_final

# Estadísticas descriptivas de los residuos
print(f"Estadísticas de los residuos:")
print(f"Media de residuos: {np.mean(residuos):.3f} (idealmente cerca de 0)")
print(f"Desviación estándar de residuos: {np.std(residuos):.3f}")
print(f"RMSE calculado de residuos: {np.sqrt(np.mean(residuos**2)):.3f}")


# Crear gráficos de diagnóstico de residuos
fig, axes = plt.subplots(2, 2, figsize=(15, 10))

# 1. Residuos vs Valores Predichos
axes[0,0].scatter(y_pred_train_final, residuos, alpha=0.6, color='blue')
axes[0,0].axhline(y=0, color='red', linestyle='--')
axes[0,0].set_xlabel('Valores Predichos')
axes[0,0].set_ylabel('Residuos')
axes[0,0].set_title('Residuos vs Valores Predichos\n(Busca patrones sistemáticos)')
axes[0,0].grid(True, alpha=0.3)

# Añadir línea de tendencia para detectar patrones
z = np.polyfit(y_pred_train, residuos, 1)
p = np.poly1d(z)
axes[0,0].plot(sorted(y_pred_train), p(sorted(y_pred_train)), "r--", alpha=0.8)

# 2. Histograma de residuos
axes[0,1].hist(residuos, bins=30, alpha=0.7, color='green', edgecolor='black')
axes[0,1].set_xlabel('Residuos')
axes[0,1].set_ylabel('Frecuencia')
axes[0,1].set_title('Distribución de Residuos\n(Debería ser aproximadamente normal)')
axes[0,1].grid(True, alpha=0.3)

# 3. Q-Q plot para normalidad
stats.probplot(residuos, dist="norm", plot=axes[1,0])
axes[1,0].set_title('Q-Q Plot de Residuos\n(Evalúa normalidad)')
axes[1,0].grid(True, alpha=0.3)

# 4. Residuos vs Valores Reales
axes[1,1].scatter(y_train, residuos, alpha=0.6, color='purple')
axes[1,1].axhline(y=0, color='red', linestyle='--')
axes[1,1].set_xlabel('Valores Reales')
axes[1,1].set_ylabel('Residuos')
axes[1,1].set_title('Residuos vs Valores Reales')
axes[1,1].grid(True, alpha=0.3)

plt.tight_layout()
plt.show()

# Test de normalidad de residuos
shapiro_stat, shapiro_p = stats.shapiro(residuos[:5000] if len(residuos) > 5000 else residuos)
print(f"\nTest de Shapiro-Wilk para normalidad de residuos:")
print(f"Estadístico: {shapiro_stat:.4f}, p-value: {shapiro_p:.4f}")
if shapiro_p > 0.05:
    print("Los residuos siguen una distribución normal (buena señal)")
else:
    print("Los residuos NO siguen una distribución normal (posible problema)")
```

### 4.3 Error por Rangos de Valores de Variable Objetivo

```python
# =============================================================================
# TERCER ANÁLISIS: EVALUACIÓN CONTEXTUAL POR RANGOS DE PUNTAJE (CON MAE)
# =============================================================================
print("\n" + "="*60)
print("3. ANÁLISIS CONTEXTUAL POR RANGOS DE PUNTAJE (MAE)")
print("="*60)

# Copiar dataset original y añadir target real y predicciones
icfes_copy = x_train_optimizado.copy()
icfes_copy['puntaje_real'] = y_train.values
icfes_copy['puntaje_pred'] = y_pred_train_final
icfes_copy['residuo'] = icfes_copy['puntaje_real'] - icfes_copy['puntaje_pred']

# Crear categorías de puntaje real
icfes_copy['categoria_puntaje'] = pd.cut(
    icfes_copy['puntaje_real'], 
    bins=5, 
    labels=['0-100', '100-200', '200-300', '300-400', '400-500']
)

# Análisis por categorías
print("Distribución por categorías de puntaje:")
print(icfes_copy['categoria_puntaje'].value_counts().sort_index())

# Calcular errores reales por categoría (MAE)
categoria_stats = []
for categoria in ['0-100', '100-200', '200-300', '300-400', '400-500']:
    mask = icfes_copy['categoria_puntaje'] == categoria
    if mask.sum() > 0:

        puntajes_categoria = icfes_copy.loc[mask, 'puntaje_real']
        residuos_categoria = icfes_copy.loc[mask, 'residuo']

        # ================================
        #   CAMBIO DE RMSE → MAE
        # ================================
        mae_categoria = np.mean(np.abs(residuos_categoria))

        categoria_stats.append({
            'Categoria': categoria,
            'N_estudiantes': mask.sum(),
            'Puntaje_promedio': puntajes_categoria.mean(),
            'MAE_categoria': mae_categoria,
            'Error_relativo (%)': (mae_categoria / puntajes_categoria.mean()) * 100
        })

df_categoria_stats = pd.DataFrame(categoria_stats)
print("\nAnálisis de error por categoría de puntaje (MAE):")
print(df_categoria_stats.to_string(index=False, float_format='%.2f'))

# ====================================================
# VISUALIZACIÓN DEL ANÁLISIS CONTEXTUAL (CON MAE)
# ====================================================
plt.figure(figsize=(15, 5))

# MAE por categoría
plt.subplot(1, 3, 1)
df_categoria_stats.plot(x='Categoria', y='MAE_categoria', kind='bar', ax=plt.gca(), color='coral')
plt.title('MAE por Categoría de Puntaje')
plt.ylabel('MAE')
plt.xticks(rotation=45)
plt.grid(True, alpha=0.3)

# Error relativo
plt.subplot(1, 3, 2)
df_categoria_stats.plot(x='Categoria', y='Error_relativo (%)', kind='bar', ax=plt.gca(), color='lightblue')
plt.title('Error Relativo por Categoría')
plt.ylabel('Error Relativo (%)')
plt.xticks(rotation=45)
plt.grid(True, alpha=0.3)

# Distribución de estudiantes
plt.subplot(1, 3, 3)
icfes_copy['categoria_puntaje'].value_counts().sort_index().plot(kind='bar', color='lightgreen')
plt.title('Distribución de Estudiantes por Categoría')
plt.ylabel('Número de Estudiantes')
plt.xticks(rotation=45)
plt.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```

## 5. Modelo Final

```python
# ============================================
# REPORTE FINAL DEL MODELO RANDOM FOREST
# ============================================

import numpy as np
import pandas as pd
from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    median_absolute_error,
)
import matplotlib.pyplot as plt

# ----------------------------------------------------
# 1. PREDICCIONES SOBRE TEST
# ----------------------------------------------------
y_pred_test = rfr_opt.predict(x_test_optimizado)

# ----------------------------------------------------
# 2. MÉTRICAS GENERALES (TEST)
# ----------------------------------------------------
mae_test = mean_absolute_error(y_test, y_pred_test)
rmse_test = np.sqrt(mean_squared_error(y_test, y_pred_test))
medae_test = median_absolute_error(y_test, y_pred_test)
r2_test = rfr_opt.score(x_test_optimizado, y_test)

print("============================================")
print(" MÉTRICAS EN TEST ")
print("============================================")
print(f"MAE   : {mae_test:.3f}   # En promedio el modelo se equivoca por ~{mae_test:.1f} puntos ICFES")
print(f"RMSE  : {rmse_test:.3f}  # Penaliza más errores grandes; indica que algunos fallos superan los {rmse_test:.1f} puntos")
print(f"MedAE : {medae_test:.3f} # La mitad de los errores son menores a ~{medae_test:.1f} puntos")
print(f"R²    : {r2_test:.4f}    # Variabilidad explicada sobre datos nuevos")
print("============================================\n")


# ----------------------------------------------------
# 3. MÉTRICAS EN TRAIN (para comparación)
# ----------------------------------------------------
y_pred_train = rfr_opt.predict(x_train_optimizado)

mae_train = mean_absolute_error(y_train, y_pred_train)
rmse_train = np.sqrt(mean_squared_error(y_train, y_pred_train))
medae_train = median_absolute_error(y_train, y_pred_train)
r2_train = rfr_opt.score(x_train_optimizado, y_train)

print("============================================")
print(" MÉTRICAS EN TRAIN ")
print("============================================")
print(f"MAE   : {mae_train:.3f}   # Error medio en puntos ICFES sobre datos vistos en entrenamiento")
print(f"RMSE  : {rmse_train:.3f}")
print(f"MedAE : {medae_train:.3f}")
print(f"R²    : {r2_train:.4f}    # Variabilidad explicada durante el entrenamiento")
print(f"OOB Score: {rfr_opt.oob_score_:.4f}")
print("============================================\n")


# ----------------------------------------------------
# 4. MAE CONTEXTUALIZADO SEGÚN LA VARIABILIDAD REAL
# ----------------------------------------------------
desv_real = y_test.std()
ratio_mae_std = mae_test / desv_real

print("============================================")
print(" MAE CONTEXTUALIZADO ")
print("============================================")
print(f"Desviación estándar real del ICFES : {desv_real:.2f}")
print(f"MAE / std                           : {ratio_mae_std:.3f}")
print(" # Indica qué proporción de la variabilidad real explica el error medio")
print("============================================\n")


# ----------------------------------------------------
# 5. MAE POR RANGOS DE 100 PUNTOS (0–500)
# ----------------------------------------------------
bins = [0, 100, 200, 300, 400, 500]
labels = ["0–100", "100–200", "200–300", "300–400", "400–500"]

df_test = x_test_optimizado.copy()
df_test["y_real"] = y_test
df_test["y_pred"] = y_pred_test

df_test["rango_icfes"] = pd.cut(
    df_test["y_real"], 
    bins=bins, 
    labels=labels,
    include_lowest=True
)

mae_por_rango = df_test.groupby("rango_icfes").apply(
    lambda df: mean_absolute_error(df["y_real"], df["y_pred"])
)

print("============================================")
print(" MAE POR RANGOS (0–500 en pasos de 100) ")
print("============================================")
print(mae_por_rango)
print("\n # Estos valores muestran cómo cambia el error")
print("   según el nivel real del puntaje ICFES.")
print("============================================\n")


# ----------------------------------------------------
# 6. GRÁFICO (opcional): MAE por rangos de puntaje
# ----------------------------------------------------
plt.figure(figsize=(8,5))
plt.bar(mae_por_rango.index.astype(str), mae_por_rango.values)
plt.title("MAE por rango de puntaje ICFES (0–500)")
plt.ylabel("MAE")
plt.xlabel("Rango del puntaje real")
plt.tight_layout()
plt.show()

```
